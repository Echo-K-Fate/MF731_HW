```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
u_cut = 1 / 11
d_cut = 1 / 9

u_max = 3 / 11
d_max = 1 / 3


# Case III: horizontal segment
sigma_u_1 = np.linspace(0, u_cut, 100)
sigma_d_1 = np.full_like(sigma_u_1, d_max)


# Case IV: diagonal segment
sigma_u_2 = np.linspace(u_cut, u_max, 200)
sigma_d_2 = (0.4 - 1.1 * sigma_u_2) / 0.9


# Case II: vertical segment
sigma_d_3 = np.linspace(0, d_cut, 100)
sigma_u_3 = np.full_like(sigma_d_3, u_max)


plt.figure(figsize=(7, 6))

plt.plot(sigma_u_1, sigma_d_1, label="Case III")
plt.plot(sigma_u_2, sigma_d_2, label="Case IV")
plt.plot(sigma_u_3, sigma_d_3, label="Case II")

plt.scatter(
    [0, u_cut, u_max, u_max],
    [d_max, d_max, d_cut, 0]
)

plt.xlabel(r"$\sigma_1^u$")
plt.ylabel(r"$\sigma_1^d$")
plt.title("Set of Calibrated Models")

plt.xlim(0, 0.32)
plt.ylim(0, 0.38)

plt.grid(True)
plt.legend()

plt.show()
```
